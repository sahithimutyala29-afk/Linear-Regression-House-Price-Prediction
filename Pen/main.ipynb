Initially start with weight and bias, because the Price = weight * area + bias.Initially the weight and bias are = 0.

```python
weight = 0
bias = 0
x_data = data["area"]
y_data = data["price"]
```

```python
def predict(x):
    input_pred = weight * x + bias
    return input_pred
```

-> y_pred = predicted value, y = actual value. The predicted and actual are diff so we have to train the computer.
If predicted value = 45, actual value=50 -> error is (predicted value-actual value). here 45-50 = -5. So the value is below the actual value. if +5 the error is above the actual value
-> Error tells us how far our prediction is from the actual answer.

```python
def error(y_pred, y):
    error = y_pred - y
    return error
```

If we have 100 houses we got 100 errors. To represent the errors we want 1 value that is "loss". 

```python
def loss(error):
    loss = np.mean(error)
    loss = loss ** 2
    return loss
```

If predictionn is wrong we know that we have to chnage Prediction. Bcoz(prediction = weight × area + bias).
How we know how to change the weight. d_weight,d_bais -> direction of change(How should the bias change to reduce the error?).

```python
def optimize(x_data, y_data, error):
    epochs = range(1000)
    global weight, bias
    learning_rate = 0.000000001
    n = len(x_data)
    for epoch in epochs:
        pred = input(x_data)
        errors = error(pred, y_data)
        d_weight = 2/n * np.sum(errors * x_data)
        d_bias = 2/n * np.sum(errors)
        weight = weight - learning_rate * d_weight
        bias = bias - learning_rate * d_bias
        loss_value = loss(errors)
        print(f"loss : {loss_value}, epoch: {epoch}")
optimize(x_data, y_data, error)
```
